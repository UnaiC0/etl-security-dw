```python
import pandas as pd
import numpy as np
print(f' Pandas: {pd.__version__}')
print(f' Numpy: {np.__version__}')

```

```python
# CELDA 2: CARGA DATASET ORIGINAL
df = pd.read_csv('data/failed_logins.csv')  
print(f' Columnas: {list(df.columns)}')
print('\n Primeras 5 filas:')
display(df.head(20))
```

```python
# Identifica nulos, duplicados y tipos datos
print('Nulos por columna:')
print(df.isnull().sum())
print(f'\nDuplicados totales: {df.duplicated().sum()}')
print('\nTipos de datos:')
print(df.dtypes)

```

```python
# Elimina NaN y crea IDs únicos perfectos
df['event_id'] = range(1, len(df) + 1)
df['event_id'] = df['event_id'].astype(int)
print(f' event_id: 1  {df["event_id"].max()} (0 NaN)')
print(df[['event_id', 'username']].head())
```

```python
df.head()
```

```python
# Quita espacios, minúsculas consistentes
text_cols = ['username', 'role', 'status', 'location', 'device', 'account_status']
for col in text_cols:
    if col in df.columns:
        df[col] = df[col].astype(str).str.strip().str.lower()

print(' Texto normalizado (lowercase + strip)')
print(df[text_cols].head())
```

```python
df.head()
```

```python
# CELDA 6: ESTANDARIZAR ROLES (3 únicos)
# USER/admin/guest → admin/user/guest
role_map = {
    'user': 'user', 'admin': 'admin', 'guest': 'guest',
    'administrator': 'admin', 'sysadmin': 'admin', 
    'root': 'admin', 'service': 'user', 'backup': 'admin'
}
df['role'] = df['role'].replace(role_map)
print(' Roles únicos (3):')
print(df['role'].value_counts())
```

```python
# CELDA 7: ESTANDARIZAR STATUS (3 únicos)
# failed/FAIL/fallido → failed/success/locked
status_map = {
    'failed': 'failed', 'fail': 'failed', 'fallido': 'failed',
    'success': 'success', 'locked': 'locked'
}
df['status'] = df['status'].replace(status_map)
print(' Status únicos (3):')
print(df['status'].value_counts())
```

```python
# CELDA 8: DEVICE → 0 NULOS
# win→Windows, ubuntu→Ubuntu, NaN→Unknown
print(f'Device NaN antes: {df["device"].isnull().sum()}')
df['device'] = df['device'].replace({'win': 'windows', 'wint': 'windows'})
df['device'] = df['device'].str.capitalize().fillna('Unknown')
print(f' Device NaN después: {df["device"].isnull().sum()}')
print(df['device'].value_counts().head())
```

```python
#LOGIN_ATTEMPTS → Unknown/Números
# "three"→3, NaN→Unknown
print(f'Login_attempts NaN antes: {df["login_attempts"].isnull().sum()}')
df['login_attempts'] = df['login_attempts'].astype(str)
df['login_attempts'] = df['login_attempts'].replace({'nan': 'Unknown', 'na': 'Unknown', 'three': '3'})
df['login_attempts'] = pd.to_numeric(df['login_attempts'], errors='coerce').fillna('Unknown')
print(' Login_attempts limpios:')
print(df['login_attempts'].value_counts(dropna=False).head())
```

```python
# CELDA 10: SOURCE_PORT → Puertos válidos
# 0-65535 válidos, resto→unknown
def clean_port(port):
    try:
        val = int(float(str(port)))
        return val if 0 <= val <= 65535 else 'unknown'
    except:
        return 'unknown'

print(f'Source_port NaN antes: {df["source_port"].isnull().sum()}')
df['source_port'] = df['source_port'].apply(clean_port).fillna('unknown')
print(' Source_port válidos:')
print(df['source_port'].value_counts().head())
```

```python
# CELDA 11: ALERT_FLAG → 0/1
# yes/no/-1/0/1 → binario 0/1
df['alert_flag'] = df['alert_flag'].astype(str).str.lower()
df['alert_flag'] = df['alert_flag'].replace({'yes': 1, 'no': 0, '-1': 0, '0': 0, '1': 1}).astype(int)
print(' Alert_flag binario:')
print(df['alert_flag'].value_counts())
```

```python
# CELDA 12: FECHAS → 99.5% parseadas
# 15+ formatos diferentes → datetime
def parse_date(date_str):
    if pd.isna(date_str): return pd.NaT
    date_str = str(date_str).strip()
    formats = ['%Y-%m-%d %H:%M:%S','%d-%m-%Y %H:%M','%Y-%m-%d','%d-%m-%Y']
    for fmt in formats:
        try: return pd.to_datetime(date_str, format=fmt)
        except: continue
    return pd.to_datetime(date_str, errors='coerce', dayfirst=True)

df['timestamp'] = df['timestamp'].apply(parse_date)
df['last_success_login'] = df['last_success_login'].apply(parse_date)
print(f' Timestamp: {df["timestamp"].notna().mean():.1%}')
print(df[['timestamp', 'last_success_login']].head())
```

```python
df.head(10)
```

```python
# CELDA 13: "Nan" LITERAL → Unknown
# Strings 'Nan'/'nan' → Unknown
nan_variants = ['Nan', 'nan', 'NaN', 'NULL', 'null']
df = df.replace(nan_variants, 'Unknown')
df = df.fillna('Unknown')

print(' "Nan" literales eliminados')
print(df[['device', 'location']].head(10))
```

```python
# CELDA 14: IPS → unknown si inválidas
# 999.999.999.999→unknown, multicast→unknown
def clean_ip(ip):
    ip_str = str(ip).strip()
    if ip_str in ['999.999.999.999', '0.0.0.0']:
        return 'unknown'
    try:
        parts = ip_str.split('.')
        nums = [int(p) for p in parts[:4]]
        if len(parts) == 4 and all(0 <= n <= 255 for n in nums):
            return ip_str
    except:
        pass
    return 'unknown'

df['ip_address'] = df['ip_address'].apply(clean_ip)
print(' IPs validadas:')
print(df['ip_address'].value_counts().head())
```

```python
# Confirma 0 nulos → dataset_clean_pandas.csv
print(f'Shape: {df.shape}')
print('Nulos por columna:')
nulls = df.isnull().sum()
print(nulls[nulls > 0] if nulls.sum() > 0 else ' no hay nulos')

# EXPORT
df.to_csv('dataset_clean_pandas.csv', index=False)
print('\n  dataset_clean_pandas.csv')
```
